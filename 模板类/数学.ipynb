## 1.1exgcd

```python

# https://www.acwing.com/solution/content/1393/这篇题解帮助理解证明

def exgcd(a, b, x, y):
    if b == 0:   # b=0时, ax + by = a, 因此x=1, y=0
        return a, 1, 0
    d, y, x = exgcd(b, a % b, y, x)
    y -= a // b * x
    return d, x, y

def gcd(a, b):
    return gcd(b, a % b) if b else a

def lcm(a, b):
    return a * b // gcd(a, b)
```

## 2.1分解质因数

```python
# 根号n的时间复杂度
def div(x):
    res = []
    i = 2
    while i * i <= x:
        if x % i == 0:
            s = 0
            while x % i == 0:
                x //= i
                s += 1
            res.append([i, s])
        i += 1
    if x > 1:                   # 最后可能还剩一个
        res.append([x, 1])
    return res
```

## 2.2阶乘分解

```python
N = int(1e6 + 10)
is_prime = [True] * N

def find(n):
    prime = []
    for i in range(2, n + 1):
        if is_prime[i]:
            prime.append(i)
        j = 0
        while i * prime[j] <= n:
            is_prime[i * prime[j]] = False
            if i % prime[j] == 0:
                break
            j += 1
    return prime

n = int(input())

prime = find(n + 1)
for i in range(2, n + 1):
    if not is_prime[i]:
        continue
    res = 0
    j = i
    while j <= n:
        res += n // j
        j *= i
    print(i, res)

```

## 2.3筛质数

```python
N = int(1e6 + 10)
is_prime = [True] * N  # 2, 3, 5, 7  质数个数数量级大概是n /（2 * lgn）
# O(nlog(logn))
def prime(n):
    for i in range(2, n + 1):
        if is_prime[i]:     # 质数的倍数一定是合数，都筛了。
            for j in range(2 * i, n + 1, i):
                is_prime[j] = False

# O(n)
def find_prime(n):
    prime = []
    for i in range(2, n + 1):
        if is_prime[i]:
            prime.append(i)
        j = 0
         # 循环不变式：pi一定是pi*x的最小质因数     t=pi*x x是t的最大因数，则pi一定不大于x的最小质因数
        while i * prime[j] <= n:
            is_prime[i * prime[j]] = False
            if i % prime[j] == 0:  # 不是最小质因数了，退出，防止重复筛
                break
            j += 1
    return len(prime)
```
